```python
# 1a, code doesn't work
songs.groupby("genre").max()["title"]
```

```python
# 1b solution
songs.sort_values("streams", ascending=False).groupby("genre")["title"].first()
```

```python
# 1b alternate solution
songs.set_index("title").groupby("genre")["streams"].idxmax()
```

```python
# 1c solution
(
    songs[songs["streams"] >= 1000000]
    .set_index("title")
    .groupby("genre")["streams"]
    .transform(lambda s: s.mean())
)
```
